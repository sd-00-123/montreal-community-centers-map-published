```python
# combine kept data columns per dataset; MTL PLACES csv

import os
import pandas as pd
import geopandas as gpd

df_places = pd.read_csv("../data/mtl_places.csv")

gdf_places = gpd.GeoDataFrame(
    df_places,
    geometry=gpd.points_from_xy(
        df_places["lon"],
        df_places["lat"]
    ),
    crs="EPSG:4326"
)

mtl_places_clean = ['titre_lieu', 'url_fiche', 'description', 
    'arrondissements', 'types', 'installations', 'activites', 'commodites', 
     'accessibilite', 'categories', 'statut_ouverture', 'horaire_par_jour', 
     'commentaires_horaire', 'reglementation', 'paiement', 'telephone', 'courriel', 
     'titre_lieu_adresse_postale', 'adresse_principale', 'adresse_secondaire', 
     'code_postal', 'ville', 'lon', 'lat', 'X', 'Y', 'geometry'
]

# Keep only rows that have the fields needed to identify and map a place
gdf_places = gdf_places.dropna(
    subset=["titre_lieu", "lon", "lat", "geometry"]
).copy()

# Fill optional text fields with blank strings
text_columns = [
    column for column in mtl_places_clean
    if column not in ["lon", "lat", "geometry"]
]

gdf_places[text_columns] = gdf_places[text_columns].fillna("")

keep = gdf_places[mtl_places_clean].copy()

output_path = "../data/mtl_places_clean_output.geojson"

keep.to_file(
    output_path,
    driver="GeoJSON",
    COORDINATE_PRECISION=6
)

print("Rows exported:", len(keep))
print(round(os.path.getsize(output_path) / 1_000_000, 1), "MB")
```

```python
# ---- settings: change these for your own data ----
FILE = "../data/stops_clean_output.geojson"
WEST, EAST, SOUTH, NORTH = -74.05, -73.40, 45.35, 45.75  # a box around your area, in degrees
EXPECTED = 8880           # how many features you expect (rows after your filters)
POPUP_FIELDS = ["stop_id", "stop_name", "stop_lon", "stop_lat", "wheelchair_boarding"]  # what you want to show in the popup
# ----------------------------------------------------

back = gpd.read_file(FILE)
problems = []
if len(back) != EXPECTED:
    problems.append(f"{len(back)} features, but you expected {EXPECTED}.")
if back.crs is None or back.crs.to_epsg() != 4326:
    problems.append(f"The file is in {back.crs}, not longitude and latitude (EPSG:4326).")
minx, miny, maxx, maxy = back.total_bounds
if not (WEST <= minx and maxx <= EAST and SOUTH <= miny and maxy <= NORTH):
    problems.append(f"Some features are outside your box: x {minx:.3f} to {maxx:.3f}, y {miny:.3f} to {maxy:.3f}.")
missing = [f for f in POPUP_FIELDS if f not in back.columns]
if missing:
    problems.append(f"Missing popup fields: {missing}.")
mb = os.path.getsize(FILE) / 1_000_000
if mb > 5:
    problems.append(f"The file is {mb:.1f} MB; over about 5 MB the page gets slow. Filter or simplify.")
print("Geometry types:", sorted(back.geom_type.unique()))
print("All checks passed" if not problems else "Not yet:\n- " + "\n- ".join(problems))
```
